```python
import os

def limpar_e_renomear_arquivos(diretorio, palavras_remover):
    # Percorre o diretório de baixo para cima (bottom-up) para evitar erros ao mudar nomes de pastas antes dos arquivos
    for raiz, pastas, arquivos in os.walk(diretorio, topdown=False):
        for arquivo in arquivos:
            nome_antigo, extensao = os.path.splitext(arquivo)
            novo_nome = nome_antigo
            
            # Remove cada palavra da lista do nome do arquivo
            for palavra in palavras_remover:
                novo_nome = novo_nome.replace(palavra, "")
            
            # Remove espaços extras que possam ter sobrado nas pontas ou duplos
            novo_nome = " ".join(novo_nome.split())
            
            # Se o nome mudou, realiza a renomeação
            if novo_nome != nome_antigo:
                # Reconstrói o nome com a extensão original
                novo_arquivo = novo_nome + extensao
                
                # Caminhos completos para o sistema operacional
                caminho_antigo = os.path.join(raiz, arquivo)
                caminho_novo = os.path.join(raiz, novo_arquivo)
                
                try:
                    os.rename(caminho_antigo, caminho_novo)
                    print(f"Renomeado: '{arquivo}' -> '{novo_arquivo}'")
                except Exception as e:
                    print(f"Erro ao renomear '{arquivo}': {e}")

# --- CONFIGURAÇÃO E EXECUÇÃO ---
if __name__ == "__main__":
    # Digite o caminho da pasta que deseja processar
    pasta_alvo = r"D:/GeoPipe/data/01_toa_images/argemiro" 
    
    # Lista de termos que você quer apagar dos nomes dos arquivos
    termos_para_apagar = ["__"]
    
    # Executa a função
    limpar_e_renomear_arquivos(pasta_alvo, termos_para_apagar)
```
