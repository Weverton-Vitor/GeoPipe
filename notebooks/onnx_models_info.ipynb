```python
import onnx
from onnx import numpy_helper

def contar_parametros_onnx(caminho_modelo: str) -> float:
    """
    Carrega um modelo ONNX e retorna a quantidade de parâmetros em milhões.
    
    Args:
        caminho_modelo (str): O caminho para o arquivo .onnx.
        
    Returns:
        float: Número total de parâmetros em milhões (M).
    """
    # Carrega o modelo ONNX
    model = onnx.load(caminho_modelo)
    
    total_params = 0
    
    # Percorre todos os inicializadores (pesos, biases) do grafo do modelo
    for initializer in model.graph.initializer:
        # Converte a matriz de parâmetros do ONNX para array do numpy e pega o tamanho
        tensor_array = numpy_helper.to_array(initializer)
        total_params += tensor_array.size
        
    # Converte para milhões (divide por 1.000.000)
    params_em_milhoes = total_params / 1_000_000
    
    return params_em_milhoes

# ==========================================
# Exemplo de uso:
# ==========================================
if __name__ == "__main__":
    path_do_seu_modelo = "D:\\GeoPipe\\data\\000_models\\deepwatermap.onnx" # Substitua pelo path correto
    
    try:
        milhoes = contar_parametros_onnx(path_do_seu_modelo)
        print(f"O modelo possui: {milhoes:.2f}M parâmetros.")
    except Exception as e:
        print(f"Erro ao ler o modelo: {e}")
```
